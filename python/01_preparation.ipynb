## Chargement des données

```python
import pandas as pd

df = pd.read_csv(
    "../raw/eco2mix-national-cons-def.csv",
    sep=";",
    keep_default_na=False
)
```

## Recherche des valeurs ND
Le fichier contient des valeurs `ND` dans deux variables : les échanges commerciaux Allemagne–Belgique et la cogénération gaz. Ces variables ne sont pas nécessaires pour répondre au besoin client et ne seront pas conservées dans le jeu de données de travail.

```python
for col in df.columns:
    if (df[col].astype(str) == "ND").any():
        print(col)
```

## Sélection des variables utiles
L'analyse porte sur l'évolution temporelle de la consommation électrique française. Seules les variables nécessaires à l'identification des observations, à leur statut et à la mesure de consommation sont conservées. Le fichier source brut reste inchangé.

```python
df_consommation = df[["Périmètre","Nature", "Date", "Heure", "Date et Heure", "Consommation (MW)"]]

df_consommation.head()
```

## Export des données

Le jeu de données contenant uniquement les variables utiles à l'analyse est exporté dans un nouveau fichier CSV destiné à être importé dans PostgreSQL. Le fichier source brut est conservé sans modification.

```python
df_consommation.to_csv(
    '../processed/eco2mix_consommation.csv',
    sep=";",
    index = False
    )
```
